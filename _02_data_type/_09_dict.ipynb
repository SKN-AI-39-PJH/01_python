# dict 
- 사전(dictionary) 자료형
- 하나의 요소(item)는 key:value 쌍으로 존재

- 같은 dict 내에서 key는 중복될 수 없다.
    - set은 순서 유지 x, dict는 가능(python 3.7이상부터)
- value는 중복 가능하다.

- key는 hashable 이어야 한다.
    - list, dict, set을 제외한 데이터 타입만 가능하다.
    - 보통 str을 제일 많이 사용

- dict는 index 대신 key를 이용해서 value을 조회한다.

````python
# dict 생성
dct = {} # 빈 dict
print(type(dct))

dct = {
    'a': 10,
    'b': 20,
    'c': 30
}

print(dct, type(dct))
````

````python
# tuple을 key로 지정
point_map = {(10, 20): 'a 좌료'}
print(point_map)

# point_map2 = {[10,20] : 'a 좌표'}
# print(point_map2) # Error
````

## dict 조회

````python
dct = {
    'a': 10,
    'b': 20,
    'c': 30
}
````

````python
# [key] 를 이용한 조회
print("dct['a']:", dct['a'])
print("dct['b']:", dct['b'])
print("dct['c']:", dct['c'])
````

````python
# dict.get(key)를 이용한 조회
print("dct.get('a'):", dct.get('a'))
print("dct.get('b'):", dct.get('b'))
print("dct.get('c'):", dct.get('c'))
````

````python
# [], get() 차이 -> 없는 key를 이용한 조회
# [key]: key가 존재하지 않으면 KeyError 발생
# print("dct['x']:", dct['x'])
````

````python
# key가 존재하지 않으면 None 반환
print("dct.get('x'):", dct.get('x'))
````

## dict 요소(item == K:V) 추가

````python
students = {
    "s001" : "홍길동",
    "s002" : "신사임당",
    "s003" : "이순신"
}

#students에 item 추가
students['s004'] = '강감찬'

# AI 답변(질문: dict.update() 메서드 사용 시 key를 따옴표로 감싸면 오류가 발생하는 이유)
# `students.update(s005='일론머스크')`에서 `s005=`는 딕셔너리 키를 직접 쓰는 문법이 아니라 **키워드 인자**
# 문법입니다. 키워드 인자의 이름은 파이썬 식별자여야 하므로 따옴표 없이 씁니다. 이때 `update()`는 그 이름을 문자열 키인
# `"s005"`로 사용합니다.
#
# 따라서 `students.update('s005'='일론머스크')`처럼 쓰면 문자열을 키워드 인자의 이름으로 쓸 수 없어 **문법 오류**가
# 발생합니다.
#
# 따옴표가 있는 키를 쓰려면 딕셔너리 형태로 전달하세요.
#
# students.update({'s005': '일론머스크'})
#
# `students['s005'] = '일론머스크'`처럼 대입해도 됩니다.
students.update(s005='일론머스크') # K=V 형식으로 작성

# update() 수행 시 dict를 이용해서 여러 item을 한 번에 추가 가능
students.update({'s006' : '빌게이츠', 's007' : '백동현'})

print(students)
````

## dict 요소(item == K:V) 제거

````python
# pop(key)을 이용한 item 제거
removed = students.pop('s001') # value만 반환된다
print(removed)
print(students)
````

## dict 내장 함수

````python
# dict() 생성 함수
dct2 = dict(a=10, b=20)
print(dct2)

dct3 = dict([('name', '신사임당'), ('age', 33)])
print(dct3)
````

````python
# dict.keys()
keys = students.keys()
print(keys, type(keys))
````

````python
# dict.values()
values = students.values()
print(values, type(values))
````

## dict unpacking

````python
# dict items unpacking
dct = dict(a=10, b=20, c=30)
print(dct)
print(dct.items())

for key, value in dct.items():
    print(key, value)
````

# dict 복사

### 값의 종류
- 원시 자료형(int, float, bool, ...)
  - 변수에 순수한 값 자체(리터럴)로 저장되는 자료형

- 참조 자료형(list, dict, set, str ...)
  - 변수에 값이 아닌 '값이 저장된 주소'를 저장하는 자료형

### 대입 
- A = B (A 변수에 B 값을 대입)
```
a = 10 # int (원시 자료형, 값 자체)
b = a  # a에 저장되어있던 10이라는 값을 복사해서 b에 대입
# a=10, b=10

lst_a = [1,2,3] # list (참조 자료형, 값이 존재하는 곳의 주소)

lst_b = lst_a  # list_a에 저장되어있던 주소를
               # list_b에 복사해서 대입

# lst_a와 lst_b가 같은 주소를 갖는다
# == 같은 저장 공간을 공유한다
```


### 얕은 복사(shallow copy)
sample = {
    'name': '기계식 키보드',
    'price': 30_000,
    'origin': 'KR',
}

# 얕은 복사 : 주소만 복사하고, 같은 저장 공간을 공유
other1 = sample  


### 깊은 복사(deep copy) : 복제의 개념
import copy
other2 = copy.deepcopy(sample)


````python
# 얕은 복사
sample = {
    'name': '기계식 키보드',
    'price': 30_000,
    'origin': 'KR',
}

other1 = sample

print("sample:", sample)
print("other1:", other1)

# sample 값 변경
sample['price'] = 35_000

print("sample:", sample)
print("other1:", other1)
````

````python
# 깊은 복사
import copy

nested_sample = {
    'name': '노트북',
    'options': ['16GB RAM', '512GB SSD']
}

other2 = copy.deepcopy(nested_sample)

print("nested_sample:",  nested_sample)
print("other2:",  other2)

# nested_sample 변경
nested_sample['name'] = '데스크탑'

print("nested_sample:",  nested_sample)
print("other2:",  other2)
````
